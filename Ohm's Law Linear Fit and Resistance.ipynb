# Ohm's Law Plotting with Linear Fitting

**Ohm's Law**- For Conductors, the Ohm's law states that the Potential Difference(V) across the ends of the conductors is directly proportional to the current(I) flowing through it.
</n> <div style="text-align:center"> $ V \propto I $  </div>

The proportionality constant is  R. So the equation becomes </n>
<div style="text-align:center"> $V = IR $

### Outline of the Program

This program reads through a Csv file having V and I data and then plots it with a linear fit to obtain the resistance from the given the slope.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('voltage_current_data.csv')
x = df['I']
y = df['V']
m,c =np.polyfit(x,y,1)
y_fit = m*x + c
plt.figure(figsize= (16,9))
plt.scatter(x,y, label= 'Data Points', color='green', s=5)
plt.plot(x, y_fit, label=f'Fit: y = {m:.2f}x + {c:.2f}', color='blue')    #.2f part ensures we get only the first two decimal points of slope and intercept
plt.xlabel('I')
plt.ylabel('V')
plt.title('Ohms Law Graph')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(fname='Ohms Law Graph')
plt.show()
```

The Resistance is given by the slope of the graph
<div style="text-align:center">
$ R = V/I $
</div>
Which is the slope of the Graph

```python
print(f'Resistance is {m:.2f} ohms')
```
