```python
import numpy as np
import matplotlib.pyplot as plt

#input parameters
u_0 = float(input('Initial speed(m/s): '))
angle = float(input('Projection angle(degrees): '))
g= 9.8

#conversion of angle to radians
theta = np.radians(angle)

#time of flight
t_f = (2*u_0*np.sin(theta))/g
#time values
t = np.linspace(0,t_f,1000)
#Position equations
x = u_0*np.cos(theta)*t
y = u_0*np.sin(theta)*t - 0.5*g*(t**2)

# Plotting
plt.figure(figsize=(16, 9))
plt.plot(x, y, label=f'u_0 = {u_0} m/s, angle = {angle}°')
plt.title('Projectile Motion Simulation')
plt.xlabel('Horizontal Distance (m)')
plt.ylabel('Vertical Distance (m)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#Complimentary angles
import numpy as np
import matplotlib.pyplot as plt

#input parameters
u_0 = float(input('Initial speed(m/s): '))
angle = float(input('Projection angle(degrees): '))
comp= 90-angle
g= 9.8

#conversion of angle to radians
theta_1 = np.radians(angle)
theta_2 = np.radians(comp)

#time of flight
t_f1 = (2*u_0*np.sin(theta_1))/g
t_f2 = (2*u_0*np.sin(theta_2))/g
#time values
t_1 = np.linspace(0,t_f1,1000)
t_2 = np.linspace(0,t_f2,1000)
#Position equations
x_1 = u_0*np.cos(theta_1)*t_1
y_1 = u_0*np.sin(theta_1)*t_1 - 0.5*g*(t_1**2)
x_2 = u_0*np.cos(theta_2)*t_2
y_2 = u_0*np.sin(theta_2)*t_2 - 0.5*g*(t_2**2)
# Plotting
plt.figure(figsize=(16, 9))
plt.plot(x_1, y_1, label=f'u_0 = {u_0} m/s, angle = {angle}°')
plt.plot(x_2, y_2, label=f'u_0 = {u_0} m/s, angle = {comp}°')
plt.title('Projectile Motion Simulation')
plt.xlabel('Horizontal Distance (m)')
plt.ylabel('Vertical Distance (m)')
plt.grid(True)
plt.legend()
plt.show()
```
